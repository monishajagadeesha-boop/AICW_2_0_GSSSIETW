```python
class Student:
    def __init__(self,usn,name,place):
        self.usn = usn
        self.name = name
        self.place = place
    def show(self):
        print(f'usn of {self.usn} having the name {self.name} is from {self.place}')
    def update(self,usn,place):
        self.usn = usn
        self.place = place
    
```

```python
stud = Student(130, "abc","mysore")
stud.name
```

```python
stud.show()
```

```python
stud.update(92,"blr")
```

```python
stud.show()
```
